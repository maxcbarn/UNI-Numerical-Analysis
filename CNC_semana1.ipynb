```python
#Em Python tem algumas sintaxes para representar números em algumas bases.

print(0b1001) #bin -> dec
print(0o157) #oct -> dec
print(0xbeba) #hex -> dec
```

```python
#Em Python, podemos usar os comandos int (truncamento) e
#a operação % (resto da divisão) para computar esta conversão da seguinte forma:
x = 9
d0 = x%2; x = int(x/2); print("d0 = %d, x = %d" % (d0,x))
d1 = x%2; x = int(x/2); print("d1 = %d, x = %d" % (d1,x))
d2 = x%2; x = int(x/2); print("d2 = %d, x = %d" % (d2,x))
d3 = x%2; x = int(x/2); print("d3 = %d, x = %d" % (d3,x))
```

```python
#Em Python, podemos computar esta conversão da parte fracionária da seguinte forma
x=0.625
d = int(2*x); x = 2*x - d; print("d = %d, x = %f" % (d,x))
d = int(2*x); x = 2*x - d; print("d = %d, x = %f" % (d,x))
d = int(2*x); x = 2*x - d; print("d = %d, x = %f" % (d,x))
```

```python
print(bin(9))
print(oct(111)) #a saída será "0o157", no Python 3 ou superior
print(hex(48826))
```

```python
#Em Python, podemos controlar a impressão de números usando o comando print. Por exemplo:
import numpy as np

print("%1.5f" % -np.pi)
print("%1.5e" % -np.pi)
```

```python
#Por truncamento
print((-0.675*1e2)/1e2)
#em notação científica normalizada:
print("%1.1e" % (int(-0.675*1e2)/1e2))
```

```python
#Por arredondamento
print("%1.2f" % (-0.675))
#em notação científica normalizada:
print("%1.1e" % (-0.675))
```
